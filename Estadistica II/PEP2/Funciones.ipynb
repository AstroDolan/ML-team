```python
import numpy as np
from scipy.stats import t
from scipy.stats import norm
```

# Función 1
## Datos agrupados

```python
def datos_agrupados(xi, fi):

    xi = np.array(xi)
    fi = np.array(fi)

    n = np.sum(fi)

    media = np.sum(xi*fi)/n

    var = (np.sum(fi*xi**2) - (np.sum(fi*xi)**2)/n)/(n-1)

    s = np.sqrt(var)

    print("--------- DATOS AGRUPADOS ---------")
    print("n =",n)
    print("Media =",round(media,4))
    print("Varianza =",round(var,4))
    print("Desviacion =",round(s,4))

    return media,var,s

    #Uso: media,var,s = datos_agrupados(xi,fi)
```

# Función 2
## Una media

```python
def prueba_media(media,
                 s,
                 n,
                 mu0,
                 alpha=0.05,
                 alternativa="!="):

    tcalc = (media-mu0)/(s/np.sqrt(n))

    gl = n-1

    if alternativa==">":

        tcrit = t.ppf(1-alpha,gl)

        pvalue = 1-t.cdf(tcalc,gl)

        decision = tcalc>tcrit

    elif alternativa=="<":

        tcrit = t.ppf(alpha,gl)

        pvalue = t.cdf(tcalc,gl)

        decision = tcalc<tcrit

    else:

        tcrit = t.ppf(1-alpha/2,gl)

        pvalue = 2*(1-t.cdf(abs(tcalc),gl))

        decision = abs(tcalc)>tcrit

    print("----------- UNA MEDIA -----------")
    print("t =",round(tcalc,4))
    print("gl =",gl)
    print("Valor crítico =",round(tcrit,4))
    print("p-value =",round(pvalue,6))

    if decision:
        print("Se rechaza H0")
    else:
        print("No se rechaza H0")
```

# Función 3
## Dos medias (Welch)

```python
def prueba_dos_medias(xbar,
                      sx2,
                      nx,
                      ybar,
                      sy2,
                      ny,
                      alpha=0.05,
                      alternativa="!="):

    tcalc = (xbar-ybar)/np.sqrt(sx2/nx + sy2/ny)

    gl = (sx2/nx + sy2/ny)**2

    gl /= ((sx2/nx)**2)/(nx-1)

    gl += ((sy2/ny)**2)/(ny-1)

    gl = (sx2/nx + sy2/ny)**2 / (((sx2/nx)**2)/(nx-1)+((sy2/ny)**2)/(ny-1))

    if alternativa==">":

        tcrit = t.ppf(1-alpha,gl)

        pvalue = 1-t.cdf(tcalc,gl)

        decision = tcalc>tcrit

    elif alternativa=="<":

        tcrit = t.ppf(alpha,gl)

        pvalue = t.cdf(tcalc,gl)

        decision = tcalc<tcrit

    else:

        tcrit = t.ppf(1-alpha/2,gl)

        pvalue = 2*(1-t.cdf(abs(tcalc),gl))

        decision = abs(tcalc)>tcrit

    print("--------- DOS MEDIAS ---------")
    print("t =",round(tcalc,4))
    print("gl =",round(gl,2))
    print("Valor crítico =",round(tcrit,4))
    print("p-value =",round(pvalue,6))

    if decision:
        print("Se rechaza H0")
    else:
        print("No se rechaza H0")
```

# Función 4
## Una proporción

```python
def prueba_proporcion(x,
                       n,
                       p0,
                       alpha=0.05,
                       alternativa="!="):

    phat = x/n

    z = (phat-p0)/np.sqrt(p0*(1-p0)/n)

    if alternativa==">":

        zcrit = norm.ppf(1-alpha)

        pvalue = 1-norm.cdf(z)

        decision = z>zcrit

    elif alternativa=="<":

        zcrit = norm.ppf(alpha)

        pvalue = norm.cdf(z)

        decision = z<zcrit

    else:

        zcrit = norm.ppf(1-alpha/2)

        pvalue = 2*(1-norm.cdf(abs(z)))

        decision = abs(z)>zcrit

    print("------- UNA PROPORCION -------")
    print("p gorro =",round(phat,4))
    print("z =",round(z,4))
    print("Valor crítico =",round(zcrit,4))
    print("p-value =",round(pvalue,6))

    if decision:
        print("Se rechaza H0")
    else:
        print("No se rechaza H0")
```

# Función 5
## Dos proporciones

```python
def prueba_dos_proporciones(x1,
                            n1,
                            x2,
                            n2,
                            alpha=0.05,
                            alternativa="!="):

    p1 = x1/n1
    p2 = x2/n2

    p = (x1+x2)/(n1+n2)

    q = 1-p

    z = (p1-p2)/np.sqrt(p*q*(1/n1+1/n2))

    if alternativa==">":

        zcrit = norm.ppf(1-alpha)

        pvalue = 1-norm.cdf(z)

        decision = z>zcrit

    elif alternativa=="<":

        zcrit = norm.ppf(alpha)

        pvalue = norm.cdf(z)

        decision = z<zcrit

    else:

        zcrit = norm.ppf(1-alpha/2)

        pvalue = 2*(1-norm.cdf(abs(z)))

        decision = abs(z)>zcrit

    print("------ DOS PROPORCIONES ------")
    print("p1 =",round(p1,4))
    print("p2 =",round(p2,4))
    print("z =",round(z,4))
    print("Valor crítico =",round(zcrit,4))
    print("p-value =",round(pvalue,6))

    if decision:
        print("Se rechaza H0")
    else:
        print("No se rechaza H0")
```

# Función 6 (extra)
## Intervalo de confianza para dos medias

```python
def ic_dos_medias(xbar,
                  sx,
                  nx,
                  ybar,
                  sy,
                  ny,
                  confianza=0.95):

    alpha = 1-confianza

    EE = np.sqrt(sx**2/nx + sy**2/ny)

    gl = (sx**2/nx + sy**2/ny)**2 / (((sx**2/nx)**2)/(nx-1)+((sy**2/ny)**2)/(ny-1))

    tcrit = t.ppf(1-alpha/2,gl)

    error = tcrit*EE

    li = (xbar-ybar)-error
    ls = (xbar-ybar)+error

    print("IC",confianza*100,"%")
    print("[",round(li,4),",",round(ls,4),"]")
```
