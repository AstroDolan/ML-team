```python
import numpy as np
from scipy.stats import t, norm
```

# Datos agrupados (sirve para cualquier tabla)

```python
# ========= DATOS AGRUPADOS =========
# Para C
# Marcas de clase
xi = np.array([25,75,125,175,250,425])

# Frecuencias
fi = np.array([2,9,11,15,3,0])

# ===================================

n = np.sum(fi)

media = np.sum(fi*xi)/n

varianza = (np.sum(fi*xi**2) - (np.sum(fi*xi)**2)/n)/(n-1)

desv = np.sqrt(varianza)

print("n =",n)
print("Media =",media)
print("Varianza =",varianza)
print("Desviacion =",desv)
```

# Prueba de hipótesis de DOS MEDIAS (Ejercicio 1)

```python
#=========================================
# PRUEBA t DE DOS MEDIAS (Welch)
#=========================================

# Datos muestra X

nx = 40
xbar = 136.875
sx2 = 3108.573

# Datos muestra Y

ny = 40
ybar = 150
sy2 = 8012.82

alpha = 0.05

#=========================================

tcalc = (ybar-xbar)/np.sqrt(sy2/ny + sx2/nx)

num = (sy2/ny + sx2/nx)**2

den = ((sy2/ny)**2)/(ny-1) + ((sx2/nx)**2)/(nx-1)

gl = num/den

tcrit = tcrit = t.ppf(1-alpha/2, gl)

pvalue = 1-t.cdf(tcalc,gl)

print("t calculado =",tcalc)
print("Grados de libertad =",gl)
print("t crítico =",tcrit)
print("p-value =",pvalue)

if tcalc>tcrit:
    print("Se rechaza H0")
else:
    print("No se rechaza H0")
```

# Prueba de UNA MEDIA (Ejercicio 2a)

```python
#=========================================
# PRUEBA t DE DOS MEDIAS (Welch)
#=========================================

# Datos muestra X

nx = 12
xbar = 272
sx2 = 1636

# Datos muestra Y

ny = 10
ybar = 335
sy2 = 1892

alpha = 0.05

#=========================================

tcalc = (ybar-xbar)/np.sqrt(sy2/ny + sx2/nx)

num = (sy2/ny + sx2/nx)**2

den = ((sy2/ny)**2)/(ny-1) + ((sx2/nx)**2)/(nx-1)

gl = num/den

tcrit = t.ppf(1-alpha,gl)

pvalue = 1-t.cdf(tcalc,gl)

print("t calculado =",tcalc)
print("Grados de libertad =",gl)
print("t crítico =",tcrit)
print("p-value =",pvalue)

if tcalc>tcrit:
    print("Se rechaza H0")
else:
    print("No se rechaza H0")
```

# Intervalo de confianza de dos medias (Ejercicio 2b)

```python
#====================================
# PRUEBA t DE UNA MEDIA
#====================================

media = 156.95
desv = 5.63
n = 41

mu0 = 155.5

alpha = 0.05

#====================================

tcalc = (media-mu0)/(desv/np.sqrt(n))

tcrit = t.ppf(1-alpha/2,n-1) #unilateral se usa t.ppf(1-alpha,n-1)

pvalue = 2*(1-t.cdf(abs(tcalc),n-1))

print("t =",tcalc)
print("t crítico =",tcrit)
print("p-value =",pvalue)

if abs(tcalc)>tcrit:
    print("Se rechaza H0")
else:
    print("No se rechaza H0")
```

# Prueba para UNA PROPORCIÓN (Ejercicio 2c)

```python
#=======================================
# IC PARA DIFERENCIA DE MEDIAS
#=======================================

xbar = 156.95
sx = 5.63
nx = 41

ybar = 154
sy = 5.2
ny = 25

confianza = 0.95

#=======================================

alpha = 1-confianza

EE = np.sqrt(sx**2/nx + sy**2/ny)

gl = (sx**2/nx + sy**2/ny)**2

gl /= ((sx**2/nx)**2)/(nx-1) + ((sy**2/ny)**2)/(ny-1)

tcrit = t.ppf(1-alpha/2,gl)

error = tcrit*EE

LI = (xbar-ybar)-error
LS = (xbar-ybar)+error

print("IC =",LI,LS)
```

# Prueba de hipótesis de dos proporciones. (Ejercicio 3)

```python
#=========================================
# PRUEBA DE DOS PROPORCIONES
#=========================================

# Datos
x1 = 15      # éxitos muestra 1
n1 = 300

x2 = 8       # éxitos muestra 2
n2 = 300

alpha = 0.02

#=========================================

p1 = x1/n1
p2 = x2/n2

# Proporción combinada
p = (x1+x2)/(n1+n2)
q = 1-p

# Estadístico Z
z = (p1-p2)/np.sqrt(p*q*(1/n1 + 1/n2))

# Valor crítico (cola derecha para H1: p1 > p2)
zcrit = norm.ppf(1-alpha)

# Valor-p
pvalue = 1 - norm.cdf(z)

print(f"p1 = {p1:.5f}")
print(f"p2 = {p2:.5f}")
print(f"p combinada = {p:.5f}")
print(f"Z = {z:.5f}")
print(f"Z crítico = {zcrit:.5f}")
print(f"p-value = {pvalue:.5f}")

if z > zcrit:
    print("Se rechaza H0")
else:
    print("No se rechaza H0")
```
